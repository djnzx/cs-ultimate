```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, ax = plt.subplots(figsize=(9, 8), facecolor='white')
ax.set_facecolor('white')

# Vectors
u = np.array([3, 1])
v = np.array([1, 3])
w = u + v          # vector addition

origin = np.zeros(2)

arrow_kw = dict(angles='xy', scale_units='xy', scale=1, width=0.00625, headwidth=5, headlength=6)

ax.quiver(*origin, *v, color='tomato',     **arrow_kw, label=r"$\mathbf{v} = (1,\,3)$")
ax.quiver(*origin, *w, color='#2ca02c',    **arrow_kw, label=r"$\mathbf{u}+\mathbf{v} = \mathbf{v}+\mathbf{u} = (4,\,4)$")

# Show parallelogram: ghost copy of v from tip of u
ax.quiver(*u, *v, color='tomato', alpha=0.35, **arrow_kw)
# Ghost copy of u from tip of v (u') — used to reach u+v
ax.quiver(*v, *u, color='royalblue', alpha=0.35, **arrow_kw,
          label=r"$\mathbf{u}' = \mathbf{u}$ translated to tip of $\mathbf{v}$")

# Scalar multiple: 2u — drawn before u so u renders on top
two_u = 2 * u
ax.quiver(*origin, *two_u, color='#9467bd', zorder=2, **arrow_kw, label=r"$2\mathbf{u} = (6,\,2)$")
ax.quiver(*origin, *u, color='royalblue',   zorder=3, **arrow_kw, label=r"$\mathbf{u} = (3,\,1)$")

# Labels
offset = 0.15
ax.text(u[0]/2 - 0.45, u[1]/2 + offset, r'$\mathbf{u}$', fontsize=14, color='royalblue', fontweight='bold')
ax.text(v[0]/2 + offset, v[1]/2,         r'$\mathbf{v}$', fontsize=14, color='tomato',    fontweight='bold')
ax.text(w[0]/2 + offset, w[1]/2,         r'$\mathbf{u}+\mathbf{v}$', fontsize=13, color='#2ca02c', fontweight='bold')
ax.text(two_u[0]/2, two_u[1]/2 - 0.35,   r'$2\mathbf{u}$', fontsize=14, color='#9467bd', fontweight='bold')
ax.text(v[0] + u[0]/2 - 0.3, v[1] + u[1]/2 + 0.1, r"$\mathbf{u}'$", fontsize=14, color='royalblue', alpha=0.6, fontweight='bold')

# Dot at origin and tips
for pt, col in [(origin,'black'), (u,'royalblue'), (v,'tomato'), (w,'#2ca02c'), (two_u,'#9467bd')]:
    ax.scatter(*pt, color=col, s=40, zorder=5)

# Grid and axes
lim = 7
ax.set_xlim(-0.5, lim)
ax.set_ylim(-0.5, lim)
ax.set_aspect('equal')
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xticks(range(lim + 1))
ax.set_yticks(range(lim + 1))
ax.grid(True, linestyle='--', alpha=0.4)
ax.set_xlabel('$x$', fontsize=13)
ax.set_ylabel('$y$', fontsize=13)
ax.set_title(r'Vectors in $\mathbb{R}^2$: Addition and Scalar Multiplication', fontsize=15, pad=14)
ax.legend(loc='upper left', fontsize=12, framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python
from manim import *
import numpy as np
from IPython.display import Video, display
from pathlib import Path

config.media_width = "75%"
config.verbosity = "WARNING"

config.pixel_height = 720
config.pixel_width = 1280
config.frame_rate = 30
config.quality = "medium_quality"

SPACE_WIDTH = 7

class HomotopyDemo(Scene):
    def construct(self):
        plane = NumberPlane(
            x_range=[-SPACE_WIDTH, SPACE_WIDTH, 1],
            y_range=[-4, 4, 1],
            background_line_style={
                "stroke_opacity": 0.5,
            },
        )

        pi_creature = Circle(
            radius=0.35,
            color=PINK,
            fill_color=PINK,
            fill_opacity=1,
        )
        pi_creature.move_to(ORIGIN)

        plane.prepare_for_nonlinear_transform()
        plane.add(pi_creature)

        def homotopy(x, y, z, t):
            norm = np.linalg.norm([x, y])
            tau = interpolate(5, -5, t) + norm / SPACE_WIDTH
            alpha = sigmoid(tau)

            return [
                x,
                y + 0.5 * np.sin(2 * np.pi * alpha),
                z,
            ]

        self.play(
            Homotopy(
                homotopy,
                plane,
                run_time=3,
            )
        )

        self.wait()


scene = HomotopyDemo()
scene.render()

video_path = scene.renderer.file_writer.movie_file_path
display(Video(video_path, embed=True))
```
